## Create dataframe data

```python
dict = {"country": ["Brazil", "Russia", "India", "China", "South Africa"],
       "capital": ["Brasilia", "Moscow", "New Dehli", "Beijing", "Pretoria"],
       "area": [8.516, 17.10, 3.286, 9.597, 1.221],
       "population": [200.4, 143.5, 1252, 1357, 52.98] }

brics = pd.DataFrame(dict)
print(brics)
```

```python
dict = {"country": ["Brazil", "Russia", "India", "China", "South Africa"],
       "capital": ["Brasilia", "Moscow", "New Dehli", "Beijing", "Pretoria"],
       "area": [8.516, 17.10, 3.286, 9.597, 1.221],
       "population": [200.4, 143.5, 1252, 1357, 52.98] }

brics = pd.DataFrame(dict, index=['a', 'b', 'c', 'd', 'e'])
print(brics)
```

## ☑️ Part 3: Subsetting with conditions

Boolean Indexing is used if user wants to filter the values of a column based on conditions from another set of columns

```python
brics['area'] >10
```

```python
brics[brics['area'] >10]
```

```python
brics[brics['country']=='Russia']
```

```python
brics[brics['country'].str.contains('ia')]
```

Multiple condition expression

```python
brics[(brics['country'].str.contains('ia')) & (brics['area'] >10)]
```

```python
brics[(brics['country']=='Russia') | (brics['country']=='India')]
```

```python
brics[brics['country'].isin(['Russia', 'India'])]
```

## ☑️ Part 2: Sorting

**Q1)** Sort the data by country

```python
# uk-500.csv has county and city rather than country
df_sorted_county = df.sort_values(by=["county", "city"], ascending=True)
df_sorted_county[["first_name", "last_name", "city", "county"]].head(10)
```

**Q2)** Sort the data by firstname and last name

```python
df_sorted_name = df.sort_values(by=["first_name", "last_name"], ascending=[True, True])
df_sorted_name[["first_name", "last_name", "email"]].head(10)
```

## ☑️ Part 3: Subsetting

**Q3)** Access the data using iloc:
- The 2nd row
- The last row
- The 1st column
- First 5 rows
- The 1st, 4th, 7th, 25th row + 1st 6th 7th columns.

```python
print("2nd row:")
print(df.iloc[1])

print("\nLast row:")
print(df.iloc[-1])

print("\n1st column:")
print(df.iloc[:, 0].head())

print("\nFirst 5 rows:")
print(df.iloc[:5])

print("\n1st, 4th, 7th, 25th row + 1st, 6th, 7th columns:")
print(df.iloc[[0, 3, 6, 24], [0, 5, 6]])
```

**Q4)** Access the data using loc:
- Select 5 first rows with all columns between 'city' and 'email’
- Display data with firstname, lastname columns

```python
print("First 5 rows, columns city to email:")
print(df.loc[0:4, "city":"email"])

print("\nFirst name and last name columns:")
print(df.loc[:, ["first_name", "last_name"]].head(10))
```

**Q5)** Retrieve data with firstname is Peter


```python
peter_df = df[df["first_name"] == "Peter"]
peter_df
```

**Q6)** Retrieve data with firstname is either Peter, Gilberto , Kaycee


```python
selected_names_df = df[df["first_name"].isin(["Peter", "Gilberto", "Kaycee"])]
selected_names_df
```

**Q7)** Find all people with email ends with gmail.com


```python
gmail_df = df[df["email"].str.endswith("gmail.com", na=False)]
gmail_df[["first_name", "last_name", "email"]].head(10)
```

**Q8)** Display data with index between 100 and 200


```python
df.loc[100:200]
```

**Q9)** Add a new column Fullname. Its values are the combination between Firstname and last name


```python
df["Fullname"] = df["first_name"] + " " + df["last_name"]
df[["first_name", "last_name", "Fullname"]].head(10)
```

**Q10)** Find the index where the fullname contains the longest characters


```python
longest_idx = df["Fullname"].str.len().idxmax()
print("Index with longest Fullname:", longest_idx)
df.loc[[longest_idx], ["first_name", "last_name", "Fullname"]]
```
